Tính đầu ra của một nơ-ron sigmoid và một mạng ReLU nhỏ giải phép XOR.

> Chạy các ô theo thứ tự từ thư mục gốc của dự án để các đường dẫn `data/...` được nhận diện. Cài các thư viện `numpy`, `pandas`, `matplotlib` và `scikit-learn` trước khi chạy.

```python
"""Tu tinh lan truyen tien qua mot neuron va mang XOR bang NumPy."""
import numpy as np
import pandas as pd


def sigmoid(z):
    return 1 / (1 + np.exp(-z))


def relu(z):
    return np.maximum(0, z)


# Mot neuron sigmoid
w = np.array([0.8, -0.5, 0.3])
b = -0.2
x = np.array([2.0, 1.0, 3.0])
z = w @ x + b
print("Mot neuron:")
print(f" z = {z:.2f}")
print(f" a = sigmoid(z) = {sigmoid(z):.4f}\n")

# Mang 2 dau vao, 2 neuron an ReLU, mot dau ra tinh XOR
W1 = np.array([[1.0, 1.0], [1.0, 1.0]])
b1 = np.array([0.0, -1.0])
W2 = np.array([1.0, -2.0])
b2 = 0.0
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=float)
H = relu(X @ W1 + b1)
Y = H @ W2 + b2
print("Mang giai XOR:")
print(pd.DataFrame({"x1": X[:, 0], "x2": X[:, 1], "h1": H[:, 0], "h2": H[:, 1], "y": Y}).to_string(index=False))
print("\nTinh ca bang mot lan:", Y)
```
