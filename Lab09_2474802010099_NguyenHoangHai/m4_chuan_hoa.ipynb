> Chạy các ô theo thứ tự từ thư mục gốc của dự án để các đường dẫn `data/...` được nhận diện. Cài các thư viện `numpy`, `pandas`, `matplotlib` và `scikit-learn` trước khi chạy.

```python
"""So sanh MLP co va khong chuan hoa tren du lieu sinh vien."""
import pandas as pd
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler


df = pd.read_csv("data/sinh_vien_mo_rong.csv")
X = df[["gio_on", "diem_giua_ky", "chi_tieu"]]
y = df["qua_mon"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

raw = MLPClassifier(hidden_layer_sizes=(20,), max_iter=3000, random_state=0).fit(X_tr, y_tr)
scaled = make_pipeline(
    StandardScaler(),
    MLPClassifier(hidden_layer_sizes=(20,), max_iter=3000, random_state=0),
).fit(X_tr, y_tr)

print(f"Doan bua ca lop qua: {y_te.mean():.4f}")
print(f"Khong chuan hoa: {accuracy_score(y_te, raw.predict(X_te)):.4f}")
print(f"Co chuan hoa: {accuracy_score(y_te, scaled.predict(X_te)):.4f}")
print(f"Khong chuan hoa, so ban du doan qua: {raw.predict(X_te).sum()} tren {len(y_te)}")
```
