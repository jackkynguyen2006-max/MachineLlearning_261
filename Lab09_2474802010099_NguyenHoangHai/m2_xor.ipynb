So sánh hồi quy logistic với MLP trên dữ liệu XOR; kèm bài tập tự tính xác suất từ trọng số đã học.

> Chạy các ô theo thứ tự từ thư mục gốc của dự án để các đường dẫn `data/...` được nhận diện. Cài các thư viện `numpy`, `pandas`, `matplotlib` và `scikit-learn` trước khi chạy.

```python
"""So sanh hoi quy logistic va MLP tren bo du lieu XOR."""
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.neural_network import MLPClassifier


df = pd.read_csv("data/xor.csv")
X = df[["x1", "x2"]]
y = df["nhan"]
X_tr, X_te, y_tr, y_te = train_test_split(
    X, y, test_size=0.25, random_state=17, stratify=y
)

lr = LogisticRegression().fit(X_tr, y_tr)
print(f"Hoi quy logistic: {accuracy_score(y_te, lr.predict(X_te)):.4f}\n")
print("So neuron an | So tham so | Accuracy train | Accuracy test")
for n_hidden in [1, 2, 4, 8]:
    model = MLPClassifier(
        hidden_layer_sizes=(n_hidden,), solver="lbfgs",
        max_iter=5000, random_state=0
    ).fit(X_tr, y_tr)
    n_params = sum(w.size for w in model.coefs_) + sum(b.size for b in model.intercepts_)
    train_acc = accuracy_score(y_tr, model.predict(X_tr))
    test_acc = accuracy_score(y_te, model.predict(X_te))
    print(f"{n_hidden:12d} | {n_params:11d} | {train_acc:14.4f} | {test_acc:.4f}")

# Bai tap 1: tu tinh xac suat lop 1 cho 5 mau dau tap test.
model = MLPClassifier(
    hidden_layer_sizes=(8,), solver="lbfgs", max_iter=5000, random_state=0
).fit(X_tr, y_tr)
X5 = X_te.iloc[:5].to_numpy()
W1, W2 = model.coefs_
b1, b2 = model.intercepts_
hidden = (X5 @ W1 + b1).clip(min=0)
manual_probability = 1 / (1 + __import__("numpy").exp(-(hidden @ W2 + b2).ravel()))
model_probability = model.predict_proba(X5)[:, 1]
print("\nBai tap 1: NumPy so voi predict_proba")
for a, b in zip(manual_probability, model_probability):
    print(f"{a:.8f} | {b:.8f} | sai khac {abs(a-b):.2e}")
```
