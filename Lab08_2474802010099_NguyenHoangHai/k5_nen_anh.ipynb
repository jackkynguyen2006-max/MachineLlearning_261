# Bài 5 — Nén màu ảnh

Thực hiện theo mục 6. Mỗi pixel RGB được xem là một điểm trong không gian ba chiều; thay pixel bằng tâm cụm tương ứng để giảm số màu.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import load_sample_image
anh=load_sample_image("china.jpg")
cao,rong,_=anh.shape
diem=anh.reshape(-1,3)/255.0
print(f"Ảnh {cao} x {rong}, {cao*rong} pixel, {len(np.unique(anh.reshape(-1,3),axis=0))} màu khác nhau")
plt.figure(figsize=(8,5)); plt.imshow(anh); plt.axis("off"); plt.title("Ảnh gốc"); plt.show()
```

## Học bảng màu và lượng hóa ảnh
Để chạy nhanh, học tâm màu trên 5.000 pixel lấy ngẫu nhiên. Sau đó dự đoán màu cho toàn bộ pixel.

```python
rng=np.random.default_rng(0)
mau=diem[rng.choice(len(diem),5000,replace=False)]
anh_nen={}
for k in [4,16,64]:
    km=KMeans(n_clusters=k,n_init=4,random_state=0).fit(mau)
    nhan=km.predict(diem)
    moi=km.cluster_centers_[nhan].reshape(cao,rong,3)
    anh_nen[k]=moi
    sai=np.abs(moi.reshape(-1,3)-diem).mean()*255
    bit=int(np.ceil(np.log2(k)))
    print(f"K={k}: {bit} bit/pixel, giảm xấp xỉ {24/bit:.1f} lần, sai lệch trung bình {sai:.2f} mức sáng")
```

```python
fig,ax=plt.subplots(1,3,figsize=(15,5))
for a,k in zip(ax,[4,16,64]):
    a.imshow(anh_nen[k]); a.set_title(f"{k} màu"); a.axis("off")
plt.tight_layout(); plt.show()
```
