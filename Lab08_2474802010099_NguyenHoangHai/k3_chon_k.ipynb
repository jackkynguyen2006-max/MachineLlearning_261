# Bài 3 — Chọn số cụm K

Thực hiện theo mục 4. So sánh phương pháp khuỷu tay (inertia) với silhouette trên dữ liệu khách hàng đã chuẩn hóa.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
df = pd.read_csv("data/khach_hang.csv")
Xs = StandardScaler().fit_transform(df[["thu_nhap", "diem_chi_tieu"]])
```

```python
ket_qua=[]
for k in range(1,11):
    km=KMeans(n_clusters=k,n_init=10,random_state=0).fit(Xs)
    sil=None if k==1 else silhouette_score(Xs,km.labels_)
    ket_qua.append({"K":k,"inertia":km.inertia_,"silhouette":sil})
bang=pd.DataFrame(ket_qua)
display(bang.round(3))
```

```python
fig, ax=plt.subplots(1,2,figsize=(12,4))
ax[0].plot(bang.K,bang.inertia,marker="o"); ax[0].set(xlabel="K",ylabel="Inertia",title="Phương pháp khuỷu tay"); ax[0].axvline(5,color="red",ls="--")
ax[1].plot(bang.K[1:],bang.silhouette[1:],marker="o"); ax[1].set(xlabel="K",ylabel="Silhouette",title="Silhouette"); ax[1].axvline(5,color="red",ls="--")
plt.tight_layout(); plt.show()
```

## Nhận xét
Inertia luôn giảm khi K tăng, vì vậy không chọn K chỉ dựa trên inertia nhỏ nhất. Quan sát điểm gập và silhouette cao nhất; với bộ dữ liệu thực hành, hai tiêu chí thường cùng gợi ý K = 5.
