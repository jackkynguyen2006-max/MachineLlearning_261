# Bài tập 2 — Chữ số viết tay

Dùng SVM trên bộ digits tích hợp của scikit-learn, in báo cáo cho 10 chữ số và xác định recall/precision thấp nhất.

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import classification_report, precision_recall_fscore_support
digits=load_digits()
X_tr,X_te,y_tr,y_te=train_test_split(digits.data,digits.target,test_size=0.25,random_state=17,stratify=digits.target)
svm=make_pipeline(StandardScaler(),SVC()).fit(X_tr,y_tr)
du=svm.predict(X_te)
print(classification_report(y_te,du,digits=4))
p,r,_,_=precision_recall_fscore_support(y_te,du,labels=list(range(10)),zero_division=0)
print("Recall thap nhat:", int(r.argmin()), float(r.min()))
print("Precision thap nhat:", int(p.argmin()), float(p.min()))
print("Macro va weighted gan nhau vi support cac chu so gan can bang.")
```

**Nhận xét:** macro avg cho mỗi chữ số trọng số bằng nhau; weighted avg tính theo số mẫu từng chữ số. Vì bộ digits có số lượng mẫu mỗi lớp khá cân bằng, hai trung bình thường gần nhau.
