# Bài tập 1 — Tự viết hàm đánh giá

Viết hàm chỉ dùng NumPy, xử lý trường hợp không có dự đoán dương tính, rồi đối chiếu với scikit-learn.

```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv(Path("../data/sinh_vien_bo_hoc.csv"))
X = df.drop(columns="bo_hoc")
y = df["bo_hoc"]
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=17, stratify=y)
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

lr = make_pipeline(StandardScaler(), LogisticRegression())
import numpy as np
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
def danh_gia(y_that, y_doan):
    yt=np.asarray(y_that); yd=np.asarray(y_doan)
    tp=np.sum((yt==1)&(yd==1)); fp=np.sum((yt==0)&(yd==1))
    fn=np.sum((yt==1)&(yd==0)); tn=np.sum((yt==0)&(yd==0))
    acc=(tp+tn)/(tp+fp+fn+tn) if tp+fp+fn+tn else 0
    p=tp/(tp+fp) if tp+fp else 0
    r=tp/(tp+fn) if tp+fn else 0
    f=2*p*r/(p+r) if p+r else 0
    return {"accuracy":acc,"precision":p,"recall":r,"f1":f}
for ten,mh in [("Logistic",lr),("Mo hinh luoi",DummyClassifier(strategy="most_frequent"))]:
    du=mh.fit(X_tr,y_tr).predict(X_te)
    print(ten, danh_gia(y_te,du))
    print("sklearn:", accuracy_score(y_te,du), precision_score(y_te,du,zero_division=0), recall_score(y_te,du,zero_division=0), f1_score(y_te,du,zero_division=0))
```

**Trả lời:** precision của mô hình lười có thể tính theo quy ước `zero_division=0`: vì không dự đoán mẫu dương tính nên TP+FP=0. Hàm xử lý bằng cách trả 0 khi mẫu số bằng 0.
