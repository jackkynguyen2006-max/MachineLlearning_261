# Bài tập 6 — MAE, RMSE và một lần đoán hỏng

Đánh giá hồi quy tuyến tính trên bộ giá nhà; cộng 5 tỷ vào một dự đoán rồi tính lại các thước đo.

```python
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
df=pd.read_csv(Path("../data/gia_nha.csv"))
X=df.drop(columns="gia"); y=df["gia"]
X_tr,X_te,y_tr,y_te=train_test_split(X,y,test_size=0.25,random_state=17)
model=LinearRegression().fit(X_tr,y_tr)
y_pred=model.predict(X_te)
def danh_gia(y,yhat):
    return {"MAE":mean_absolute_error(y,yhat),"RMSE":np.sqrt(mean_squared_error(y,yhat)),"R2":r2_score(y,yhat)}
print("Ban dau:",danh_gia(y_te,y_pred))
y_hong=y_pred.copy(); idx=0; y_hong[idx]+=5
print(f"Sau khi cong 5 ty vao du doan mau {idx}:",danh_gia(y_te,y_hong))
print("Sai so lon bi RMSE phat nang hon MAE do binh phuong sai so truoc khi lay trung binh.")
```

**Nhận xét:** MAE thay đổi theo sai số tuyệt đối trung bình; RMSE nhạy hơn với một sai số lớn vì bình phương sai số; R² có thể giảm mạnh khi tổng bình phương sai số tăng.
