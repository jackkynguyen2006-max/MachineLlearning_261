# Bài tập 5 — Cân bằng lớp cho Random Forest và SVC

So sánh precision, recall, F1 trước/sau class_weight=balanced trên tập kiểm tra.

```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv(Path("../data/sinh_vien_bo_hoc.csv"))
X = df.drop(columns="bo_hoc")
y = df["bo_hoc"]
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.25, random_state=17, stratify=y)
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import precision_score,recall_score,f1_score
models={"RF thuong":RandomForestClassifier(n_estimators=200,random_state=0),"RF can bang":RandomForestClassifier(n_estimators=200,random_state=0,class_weight="balanced"),"SVC thuong":make_pipeline(StandardScaler(),SVC(probability=True,random_state=0)),"SVC can bang":make_pipeline(StandardScaler(),SVC(probability=True,random_state=0,class_weight="balanced"))}
for ten,mh in models.items():
    du=mh.fit(X_tr,y_tr).predict(X_te)
    print(f"{ten:14s} precision={precision_score(y_te,du,zero_division=0):.4f} recall={recall_score(y_te,du,zero_division=0):.4f} F1={f1_score(y_te,du,zero_division=0):.4f}")
```

**Nhận xét:** so sánh mức thay đổi từng chỉ số của từng mô hình để kết luận mô hình nào thay đổi nhiều nhất.
