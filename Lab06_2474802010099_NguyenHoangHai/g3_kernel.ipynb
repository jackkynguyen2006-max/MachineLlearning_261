## 4. Nâng chiều bằng tay và kernel

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

x=np.array([-4.,-3.4,-2.8,-2.2,-1.2,-.7,-.2,.3,.8,1.3,2.3,2.9,3.5,4.])
y=(np.abs(x)<1.8).astype(int)
best=max(np.mean((x<=t).astype(int)==y) for t in x)
print(f'Ngưỡng đơn trên x chính xác tối đa: {best:.4f}')
X_lift=np.c_[x,x**2]
svm_lift=SVC(kernel='linear',C=1000).fit(X_lift,y)
print(f'Thêm x², SVM tuyến tính chính xác: {svm_lift.score(X_lift,y):.4f}')
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

def doc_csv(ten):
    ung_vien = [Path('data') / ten, Path(ten)]
    if ten == 'sinh_vien.csv':
        ung_vien += [Path('sinh_vien (1).csv'), Path('data/sinh_vien (1).csv')]
    for p in ung_vien:
        if p.exists():
            print(f'Đọc dữ liệu: {p}')
            return pd.read_csv(p)
    raise FileNotFoundError(f'Không tìm thấy {ten}. Đã tìm: {ung_vien}')
```

```python
from sklearn.metrics import accuracy_score
df=doc_csv('hai_vang_trang.csv')
X=df[['x1','x2']]; y=df['nhan']
X_tr,X_te,y_tr,y_te=train_test_split(X,y,test_size=.25,random_state=17,stratify=y)
models={'SVM tuyến tính':SVC(kernel='linear'),'SVM đa thức bậc 3':SVC(kernel='poly',degree=3),'SVM RBF':SVC(kernel='rbf'),'Hồi quy logistic':LogisticRegression(),'KNN k=13':KNeighborsClassifier(n_neighbors=13)}
for name,m in models.items():
    m.fit(X_tr,y_tr)
    print(f'{name:24s}: {accuracy_score(y_te,m.predict(X_te)):.4f}')
print('Support vectors RBF:',len(models['SVM RBF'].support_),'trên',len(X_tr))
```

**Nhận xét:** Dữ liệu một chiều không tách được bằng một ngưỡng, nhưng biểu diễn thêm x² giúp một đường thẳng phân lớp. Kernel thực hiện phép biến đổi tương tự mà không cần tạo tường minh các cột mới. RBF có thể tạo ranh giới cong.
