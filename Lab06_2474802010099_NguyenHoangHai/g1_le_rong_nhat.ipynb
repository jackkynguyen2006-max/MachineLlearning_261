## 2. Đường có lề rộng nhất và vectơ hỗ trợ

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.svm import SVC

X, y = make_blobs(n_samples=24, centers=[[2, 2], [6, 6]],
                  cluster_std=0.9, random_state=0)
y = 1 - y
svm = SVC(kernel='linear', C=1000).fit(X, y)
w, b = svm.coef_[0], svm.intercept_[0]
norm_w = np.linalg.norm(w)
print(f'w1={w[0]:.3f}, w2={w[1]:.3f}, b={b:.3f}')
print(f'Độ rộng lề = 2/||w|| = {2/norm_w:.4f}')
print('Số vectơ hỗ trợ:', len(svm.support_))
for i in svm.support_:
    print(f'điểm {i:2d}: ({X[i,0]:.2f}, {X[i,1]:.2f}), lớp={y[i]}, f(x)={svm.decision_function(X[i:i+1])[0]:+.3f}')
ngoai = np.setdiff1d(np.arange(len(X)), svm.support_)
print(f'|f(x)| nhỏ nhất ở điểm không hỗ trợ: {np.min(np.abs(svm.decision_function(X[ngoai]))):.3f}')
sv_only = SVC(kernel='linear', C=1000).fit(X[svm.support_], y[svm.support_])
print('Học chỉ từ support vectors:', sv_only.coef_[0], sv_only.intercept_[0])
moi=np.array([[3.,3.],[5.,4.5]])
for p,f,c in zip(moi,svm.decision_function(moi),svm.predict(moi)):
    print(f'Điểm {tuple(p)}: f={f:+.3f}, dự đoán lớp {c}')
```

```python
fig, ax = plt.subplots(figsize=(7,5))
ax.scatter(X[:,0],X[:,1],c=y,cmap='coolwarm',edgecolor='k')
ax.scatter(X[svm.support_,0],X[svm.support_,1],s=180,facecolors='none',edgecolors='gold',linewidths=2,label='Support vectors')
xx=np.linspace(X[:,0].min()-1,X[:,0].max()+1,200)
for level,style in [(0,'-'),(-1,'--'),(1,'--')]:
    if abs(w[1])>1e-12: ax.plot(xx,(level-b-w[0]*xx)/w[1],style,color='black')
ax.set(xlabel='x1',ylabel='x2',title='SVM tuyến tính: ranh giới và lề'); ax.legend(); ax.grid(alpha=.2)
```

**Nhận xét:** Chỉ các điểm sát lề (support vectors) quyết định ranh giới. Giá trị decision function biểu thị phía và độ xa theo chuẩn của hàm, không phải xác suất. Điểm có |f(x)| < 1 nằm trong lề.
