## 6. Nhận dạng chữ số viết tay

```python
import numpy as np
from sklearn.datasets import load_digits
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

X,y=load_digits(return_X_y=True)
print('Kích thước:',X.shape,'| cường độ:',X.min(),'-',X.max())
print('Số ảnh từng chữ số:',np.bincount(y))
X_tr,X_te,y_tr,y_te=train_test_split(X,y,test_size=.25,random_state=17,stratify=y)
svm=SVC(kernel='rbf').fit(X_tr,y_tr)
pred=svm.predict(X_te)
print(f'SVM RBF accuracy={accuracy_score(y_te,pred):.4f}; support vectors={len(svm.support_)} / {len(X_tr)}')
wrong=np.flatnonzero(pred!=y_te)
print(f'Đoán sai {len(wrong)} / {len(y_te)} ảnh')
for i in wrong: print(f'ảnh test {i}: thật={y_te[i]}, đoán={pred[i]}')
```

```python
models={'SVM có chuẩn hoá':make_pipeline(StandardScaler(),SVC()),'KNN k=5':KNeighborsClassifier(),'Logistic có chuẩn hoá':make_pipeline(StandardScaler(),LogisticRegression(max_iter=2000))}
for name,m in models.items():
    m.fit(X_tr,y_tr)
    print(f'{name:24s}: {accuracy_score(y_te,m.predict(X_te)):.4f}')
```
